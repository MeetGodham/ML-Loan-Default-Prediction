```python
newdf = df
```

```python
numeric_cols = newdf.drop("Default", axis=1).select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
    Q1 = newdf[col].quantile(0.25)
    Q3 = newdf[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    median = newdf[col].median()

    newdf.loc[newdf[col] < lower, col] = median
    newdf.loc[newdf[col] > upper, col] = median
```
